```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

path = "final_model"   # путь к папке, поправьте под своё расположение
tok = AutoTokenizer.from_pretrained(path)
mdl = AutoModelForSequenceClassification.from_pretrained(path)
mdl.eval()

texts = ["Не плохо, но могло быть лучше",
         "Не ожидала, что будет так хорошо",
         "Размер не подошёл, возвращаю",
         "Платье красивое, но села после стирки"]

enc = tok(texts, truncation=True, max_length=128, padding=True, return_tensors="pt")
with torch.no_grad():
    print(mdl(**enc).logits.argmax(dim=-1).numpy())
```
